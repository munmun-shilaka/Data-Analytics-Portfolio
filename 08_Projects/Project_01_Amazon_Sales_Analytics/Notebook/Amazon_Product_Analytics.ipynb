```python
import pandas as pd
import matplotlib.pyplot as plt
```

# 📂 Load Dataset

The dataset is loaded using the Pandas library. The first five records are displayed to verify that the data has been imported successfully.

```python
df = pd.read_csv("08_Projects/Project_01_Amazon_Sales_Analytics/Dataset/amazon.csv")

df.head()
```

# 🧹 Data Cleaning

In this step, the dataset is inspected for missing values and cleaned by removing incomplete records to improve data quality before analysis.

```python
print("Dataset Shape Before Cleaning:")
print(df.shape)

print("\nMissing Values:")
print(df.isnull().sum())

df = df.dropna()

print("\nDataset Shape After Cleaning:")
print(df.shape)

print("\nMissing Values After Cleaning:")
print(df.isnull().sum())
```

# 📊 Exploratory Data Analysis (EDA)

Exploratory Data Analysis is performed to understand the structure of the dataset, identify important statistics, and explore product categories.

```python
print("Dataset Information")
df.info()

print("\nStatistical Summary")
display(df.describe())

print("\nUnique Categories")
print(df["category"].nunique())

print("\nTop 10 Categories")
display(df["category"].value_counts().head(10))
```

# 📈 Business Analysis

Business analysis is performed to identify the most valuable insights from the dataset, including the highest-rated products, top discounts, pricing trends, and product category performance.

```python
# Highest Rated Products
print("Top 10 Highest Rated Products")
display(
    df[["product_name", "rating"]]
    .sort_values(by="rating", ascending=False)
    .head(10)
)

# Most Expensive Product
print("\nMost Expensive Product")
display(
    df[["product_name", "actual_price"]]
    .sort_values(by="actual_price", ascending=False)
    .head(1)
)

# Top Categories
print("\nTop 5 Categories")
display(df["category"].value_counts().head())

# Rating Distribution
print("\nRating Distribution")
display(df["rating_category"].value_counts())

# Average Price
print("\nAverage Price")
print(df["actual_price"].mean())
```

```python
print(df.columns)
```

```python
# Convert rating to numeric
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")

# Create Rating Category
df["rating_category"] = pd.cut(
    df["rating"],
    bins=[0, 2, 3.5, 4.5, 5],
    labels=["Poor", "Average", "Good", "Excellent"]
)

print(df["rating_category"].value_counts())
```
